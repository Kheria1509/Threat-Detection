```python
import cv2
import numpy as np

# Load YOLOv8
net = cv2.dnn.readNet('yolov8.weights', 'yolov8.cfg')  # Change paths if necessary
classes = []
with open('coco.names', 'r') as f:
    classes = f.read().strip().split('\n')
layer_names = net.getUnconnectedOutLayersNames()

# Open webcam
cap = cv2.VideoCapture(0)  # 0 corresponds to the default webcam

while True:
    # Capture frame-by-frame
    ret, frame = cap.read()

    # YOLOv8 inference
    blob = cv2.dnn.blobFromImage(frame, 1 / 255.0, (416, 416), swapRB=True, crop=False)
    net.setInput(blob)
    outputs = net.forward(layer_names)

    # Process detection results
    for output in outputs:
        for detection in output:
            scores = detection[5:]
            class_id = np.argmax(scores)
            confidence = scores[class_id]
            if confidence > 0.5:
                # Object detected, you can customize the drawing code here
                # Example: cv2.rectangle(frame, (x, y), (x + w, y + h), (0, 255, 0), 2)

    # Display the resulting frame
    cv2.imshow('YOLOv8 Webcam', frame)

    # Break the loop if 'q' key is pressed
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Release the webcam and close all windows
cap.release()
cv2.destroyAllWindows()
```
